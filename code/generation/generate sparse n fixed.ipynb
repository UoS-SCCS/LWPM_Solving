```python
from sage.all import *
from sage.numerical.mip import MIPSolverException
import random as pyrandom


def sparse_coeffs_GF2(t, weightP, seed=None):

    if weightP < 1:
        raise ValueError("weight must be >= 1.")

    if weightP > t + 1:
        raise ValueError("weight > t!")

    F = GF(2)
    R.<x> = PolynomialRing(F)

    rng = pyrandom.Random(int(seed) if seed is not None else None)

    # Force degree exactly t
    support = [t]

    # Choose remaining weightP - 1 positions from 0,...,t-1
    if weightP > 1:
        support += rng.sample(range(t), weightP - 1)

    coeffs = [0] * (t + 1)
    for i in support:
        coeffs[i] = 1

    P = R(coeffs)

    return P, coeffs, sorted(support)
```

```python
from random import seed
import os, json, csv


out_dir = r""

os.makedirs(out_dir, exist_ok=True)

N = 250
t, d, w, weightP = 400, 100, 10, 10

for i in range(N):
    print(f"Iteration {i+1}/{N}")

    with \
        open(os.path.join(out_dir, "p.txt"), "a+", encoding="utf-8") as f_P, \
        open(os.path.join(out_dir, "coeffs_p.txt"), "a+", encoding="utf-8") as f_c, \
        open(os.path.join(out_dir, "support_p.txt"), "a+", encoding="utf-8") as f_s:
        
        P, coeffs, support = sparse_coeffs_GF2(t=t,weightP=weightP)
                
        print(str(P))
        
        f_P.write(str(P) + "\n")
        f_c.write(str(coeffs) + "\n")
        f_s.write(str(support) + "\n")
        
    
```
